import CSV data file

```python
import pandas as pd
dataset = pd.read_csv('/content/drive/MyDrive/FFNC_iris_Dataset/iris.csv').values
print(dataset.shape)
print(dataset)
```

Separate Features and Label

Separate dataset into data

```python
data = dataset[:,0:4]
print(data)
```

separate dataset into tareget

```python
target = dataset[:,4]
print(target)
```

### Create Model Using 2 Libraries {Tensorflow and Keras}

```python
from keras.models import Sequential
from keras.layers import Dense #Fully-connected Layer

model = Sequential() #empty neural network
model.add(Dense(16,input_dim = 4,activation='relu')) #Layer 1 first Hidden layer
model.add(Dense(40,activation='relu')) #adding layer 2 (2nd Hidden layer)
model.add(Dense(10,activation='relu')) #adding layer 3 (3rd Hidden layer)
model.add(Dense(3,activation='softmax')) #Output layer

model.compile(loss = 'categorical_crossentropy',optimizer = 'adam', metrics = ['accuracy'])

model.summary()
```

Convert category



```python
from keras.utils import to_categorical
new_target = to_categorical(target)
```

```python
print(new_target)
```

# Validation{Performance Check}

Dataset divide 2 part - {train / validation}

```python
from sklearn.model_selection import train_test_split

train_data,test_data, train_target, test_target = train_test_split(data,new_target,test_size = 0.2)
```

```python
print(train_data.shape)
print(test_data.shape)
```

```python
print(test_data)
```

```python
print(test_target)
```

# Train the Model

train_data = 120 / batch_sise = 4
/step = 120/4 = 30 / {mini-natch gadient}

```python
model.fit(train_data,train_target, epochs=200, validation_data=(test_data,test_target),batch_size = 4) # train the Nueral Network{batch size default 32//}
```

Draw Special Graph

```python
from matplotlib import pyplot as plt

plt.plot(model.history.history['loss'],label = 'loss')
plt.plot(model.history.history['val_loss'],label = 'val_loss')
plt.legend()
plt.xlabel('# epoch')
plt.ylabel('loss')
```

```python
plt.plot(model.history.history['accuracy'], label = 'accuracy')
plt.plot(model.history.history['val_accuracy'], label = 'val_accuracy')
plt.legend()
plt.xlabel('# epoch')
plt.ylabel('accuracy')
```
